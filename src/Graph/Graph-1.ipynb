```python
def fromListOfVertexToIncedent(vertex, edges):
    matrIncedent = [[0] * len(vertex) for _ in range(len(edges))]
    
    
    v_idx = {name: i for i, name in enumerate(vertex)}
    
    for rebro_idx, (v1, v2, dlinaRebra) in enumerate(edges):
        matrIncedent[rebro_idx][v_idx[v1]] = dlinaRebra
        matrIncedent[rebro_idx][v_idx[v2]] = dlinaRebra
        
    return matrIncedent


vershini = ['A', 'B', 'C', 'D']
rebro = [('A', 'B', 3), ('A', 'D', 2), ('B', 'C', 4)]

matrixIncedent = fromListOfVertexToIncedent(vershini, rebro)

for i in matrixIncedent:
    print(i)
```

```python
def fromListOfVertexToSootv(vertex, edges):
    matrSootv = [[0] * len(vertex) for _ in range(len(vertex))]

    v_idx = {name: i for i, name in enumerate(vertex)}

    for rebro_idx, (v1, v2, dlinaRebra, flag) in enumerate(edges):
        matrSootv[v_idx[v1]][v_idx[v2]] = dlinaRebra
        matrSootv[v_idx[v2]][v_idx[v1]] = dlinaRebra * flag

    return matrSootv


vershini = ['A', 'B', 'C', 'D']
rebro = [('A', 'B', 3, 1), ('A', 'D', 2, 1), ('B', 'C', 4, -1)]

matrixSootv = fromListOfVertexToSootv(vershini, rebro)

for i in matrixSootv:
    print(i)
```
